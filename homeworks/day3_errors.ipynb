# Day 3 extra homework: recognize a deliberate error and fix the input

## My prediction (written before running)

| Input | What I expect |
|---|---|
| `None` | `ValueError`: Enter a non-empty question string. |
| Blank spaces | `ValueError`: Enter a non-empty question string. |
| 501 characters | `ValueError`: Use at most 500 characters. |

```python
import json


def validate_public_question(question):
    if not isinstance(question, str) or not question.strip():
        raise ValueError('Enter a non-empty question string.')
    if len(question) > 500:
        raise ValueError('Use at most 500 characters.')
    return question.strip()
```

```python
cases = [(None, 'Enter a non-empty question string.'),
         ('   ', 'Enter a non-empty question string.'),
         ('x' * 501, 'Use at most 500 characters.')]

rows = []
for value, expected in cases:
    try:
        validate_public_question(value)
    except ValueError as error:
        print(type(error).__name__ + ':', error)
        assert str(error) == expected
        rows.append({'error_type': type(error).__name__, 'message': str(error), 'expected_rejection': True})
    else:
        raise AssertionError('Invalid input was accepted.')
```

```python
corrected = ' How can I check my application status? '
assert validate_public_question(corrected) == 'How can I check my application status?'

result = {'intentional_caught_errors': rows, 'corrected_input': 'accepted', 'checks': 'passed'}
print(json.dumps(result, indent=2))
```

## Explanation

- `raise` stops the function and signals a problem with an exception. Here `validate_public_question` raises `ValueError` when the input is not a non-empty string, or is longer than 500 characters.
- `try` runs the code that might fail, here the call to the validator.
- `except ValueError as error` catches that exception, so the notebook keeps running instead of stopping with a traceback, and the message is available as `str(error)`.

**Why a caught rejection is a successful test:** the three inputs are invalid on purpose. The validator is correct when it rejects them with the exact expected message, and the `assert` checks that. If it had accepted one of them, the `else` branch would raise `AssertionError` and the test would fail. The corrected question is trimmed and accepted, which shows the fix.
